```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import f1_score
```

```python
dataset=pd.read_csv('preprocessedEarthquake.csv')
df=dataset
indep=df[['magnitude','sig','depth']]
dep=df[['alert']]
```

```python
x_train,x_test,y_train,y_test=train_test_split(indep,dep,test_size=0.25,random_state=0)
```

```python
DTC=DecisionTreeClassifier(criterion='gini',max_features='sqrt',random_state=0,splitter='best')
DTC.fit(x_train,y_train)
```

```python
y_pred=DTC.predict(x_test)
```

```python
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
report=classification_report(y_test,y_pred)
print(report)
```

```python
F1=f1_score(y_test,y_pred,average='weighted')
F1
```

```python
import pickle
file_name='EarthQuake_Alert_Prediction.sav'
pickle.dump(DTC,open(file_name,'wb'))
```
